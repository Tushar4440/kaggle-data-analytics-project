## Q14 Find month over month growth comparison for 2022 & 2023 sales eg: jan 2022 vs jan 2023

```python
df1.head(2)
```

```python
df1['Year'].unique()
```

```python
a = df1.groupby(['Month','Year']).sum('Total Sales')['Total Sales'].reset_index()
```

```python
ap = a.pivot_table(index='Month', columns='Year', values='Total Sales', aggfunc = 'sum')
ap
```

```python
ap['YOY_growth']= ap[2022] - ap[2023]
ap
```

## Q16 Which sub category had highest growth by profit in 2023 compare to 2022

```python
df1.head(2)
```

```python
m = df1.groupby(['Sub Category','Year']).sum('Total Profit')['Total Profit'].reset_index()
m.head()
```

```python
n = m.pivot_table(index='Sub Category', columns='Year', values='Total Profit')
n.head()
```

```python
n['diff'] = n[2023] - n[2022]
n.sort_values(by='diff', ascending=False).head(1)
```
